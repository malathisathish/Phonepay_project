```python
# Path to the aggregated insurance data
Agg_ins_state_path=r"C:\Users\sathishkumar\Downloads\pulse\data\aggregated\insurance\country\india\state"
Agg_state_list=os.listdir(Agg_ins_state_path)

#initialize columns for the dataframe
agg_ins_data={'State':[],'Year':[],'Quarter':[],'Transaction_type':[],'Transaction_count':[],'Transaction_amount':[]}

#using for loop to extract the path
for state in Agg_state_list:
  years=os.listdir(f"{Agg_ins_state_path}/{state}")
  for year in years:
    quarter=os.listdir(f"{Agg_ins_state_path}/{state}/{year}")
    for qtr in quarter:
      # Check if the entry is a json file
      if qtr.endswith('.json'):
        path=f"{Agg_ins_state_path}/{state}/{year}/{qtr}"
        print(f"Processing file: {path}")

        #loding json file
        try:
          with open(path,'r') as f:
             data=json.load(f)
          #print(data)

          #extracting data
          for i in data['data']['transactionData']:
             insurance=i['name']
             count=i['paymentInstruments'][0]['count']
             amount=i['paymentInstruments'][0]['amount']
             print(insurance,count,amount,state,year,qtr)

             agg_ins_data['State'].append(state)
             agg_ins_data['Year'].append(year)
             agg_ins_data['Quarter'].append(int(qtr.strip('.json')))
             agg_ins_data['Transaction_type'].append(insurance)
             agg_ins_data['Transaction_count'].append(count)
             agg_ins_data['Transaction_amount'].append(amount)

        except KeyError:
            print(f"Skipping file due to missing keys: {path}")
        except FileNotFoundError:
            print(f"Skipping file not found: {path}")
        except Exception as e:
            print(f"An error occurred while processing {path}: {e}")

#creating dataframe
agg_ins_df=pd.DataFrame(agg_ins_data)

agg_ins_df["State"]=agg_ins_df["State"].str.replace('andaman-&-nicobar-islands','Andaman & Nicobar Islands')
agg_ins_df["State"]=agg_ins_df["State"].str.replace("-"," ")
agg_ins_df["State"]=agg_ins_df["State"].str.title()
agg_ins_df["State"]=agg_ins_df["State"].str.replace('Dadra & Nagar Haveli & Daman & Diu', 'Dadra and Nagar Haveli and Daman and Diu')

agg_ins_df
```

```python
#3.3 State-District Quarterly Insurance Growth Rate
#Insight:Ideal for targeted marketing, partnerships, and investment decisions in rapidly growing regions.
query="""WITH previous_quarter AS (
    SELECT 
        "State", 
        "District", 
        CAST("Year" AS INT) AS "Year", 
        "Quarter", 
        "Transaction_count" AS prev_count
    FROM map_ins_table
),
current_quarter AS (
    SELECT 
        "State", 
        "District", 
        CAST("Year" AS INT) AS "Year", 
        "Quarter", 
        "Transaction_count" AS curr_count
    FROM map_ins_table
)
SELECT 
    curr."State",
    curr."District",
    curr."Year",
    curr."Quarter",
    prev.prev_count,
    curr.curr_count,
    ROUND(((curr.curr_count - prev.prev_count) * 100.0) / NULLIF(prev.prev_count, 0), 2) AS Growth_Percentage
FROM current_quarter curr
JOIN previous_quarter prev
  ON curr."State" = prev."State"
  AND curr."District" = prev."District"
  AND curr."Year" = prev."Year"
  AND curr."Quarter" = prev."Quarter" + 1
WHERE curr."Year" >= 2022  -- Now works correctly after casting
ORDER BY Growth_Percentage DESC
LIMIT 20;"""
insurance_growth_rate_df = pd.read_sql(query, engine)
insurance_growth_rate_df
```

```python
#3.4state,district,quater wise insurance growth trajectory
#Insight: Analyzes insurance growth trends over time in each state and district, helping to identify successful markets and areas needing improvement.
query="""SELECT 
    curr."State",
    curr."District",
    curr."Year",
    curr."Quarter",
    curr."Transaction_count" AS Current_Transaction_Count,
    prev."Transaction_count" AS Previous_Transaction_Count,
    ROUND(
        ((curr."Transaction_count" - prev."Transaction_count") * 100.0) / NULLIF(prev."Transaction_count", 0), 2
    ) AS Growth_Percentage
FROM map_ins_table curr
JOIN map_ins_table prev
  ON curr."State" = prev."State"
 AND curr."District" = prev."District"
 AND curr."Year" = prev."Year"
 AND curr."Quarter" = prev."Quarter" + 1
ORDER BY Growth_Percentage DESC;"""
insurance_growth_trajectory_df = pd.read_sql(query, engine)
insurance_growth_trajectory_df
```

```python
#3.5 High-Priority Regions for Insurance Growth
#Insight: Identifies regions with high potential for insurance growth based on transaction volume and user engagement.
query="""SELECT 
    "State", 
    "District",
    SUM("Transaction_count") AS Total_Insurance_Transactions,
    COUNT(DISTINCT CONCAT("Year", '-', "Quarter")) AS Active_Quarters,
    ROUND(SUM("Transaction_count") * 1.0 / COUNT(DISTINCT CONCAT("Year", '-', "Quarter")), 2) AS Avg_Transactions_Per_Quarter
FROM map_ins_table
GROUP BY "State", "District"
HAVING SUM("Transaction_count") > 10000  
ORDER BY Avg_Transactions_Per_Quarter DESC
LIMIT 15;"""
high_priority_insurance_df = pd.read_sql(query, engine)
high_priority_insurance_df
```

```python
#5.3  Districts with High Users but Low Engagement
#Insight: Identifies districts with high user counts but low engagement, indicating potential issues with user retention or app usability.
query="""SELECT 
    "State", "District", "Year", "Quarter",
    SUM("RegisteredUsers") AS Total_Users,
    SUM("AppOpens") AS Total_App_Opens,
    ROUND(SUM("AppOpens")::numeric / NULLIF(SUM("RegisteredUsers"), 0), 2) AS Engagement_Rate
FROM map_user_table
GROUP BY "State", "District", "Year", "Quarter"
HAVING SUM("RegisteredUsers") > 1000 AND ROUND(SUM("AppOpens")::numeric / NULLIF(SUM("RegisteredUsers"), 0), 2) < 1
ORDER BY Engagement_Rate ASC;"""
low_engagement_districts_df = pd.read_sql(query, engine)
low_engagement_districts_df
```

```python
#5.4 States with Fastest Growth in Registrations (Quarter-over-Quarter)
#Insight: Identifies states with the highest growth rates in user registrations, indicating successful user acquisition strategies.
query="""WITH prev_quarter AS (
    SELECT "State", "Year", "Quarter", SUM("RegisteredUsers") AS reg_prev
    FROM map_user_table
    GROUP BY "State", "Year", "Quarter"
),
curr_quarter AS (
    SELECT "State", "Year", "Quarter", SUM("RegisteredUsers") AS reg_curr
    FROM map_user_table
    GROUP BY "State", "Year", "Quarter"
)
SELECT 
    curr."State",
    curr."Year",
    curr."Quarter",
    ROUND(
        ((curr.reg_curr - prev.reg_prev)::numeric / NULLIF(prev.reg_prev, 0)) * 100, 2
    ) AS Growth_Percentage
FROM curr_quarter curr
JOIN prev_quarter prev
  ON curr."State" = prev."State"
  AND curr."Year" = prev."Year"
  AND curr."Quarter" = prev."Quarter" + 1
ORDER BY Growth_Percentage DESC
LIMIT 10;"""
fastest_growth_registrations_df = pd.read_sql(query, engine)
fastest_growth_registrations_df
```

```python
#5.5 Most Popular Device Brands (All India)
#Insight: Identifies the most popular device brands among users across India, helping to inform marketing and product strategies.
query="""SELECT 
    "Device_Brand",
    SUM("Transaction_count") AS Total_Users,
    ROUND(AVG("Percentage"):: numeric, 2) AS Avg_Usage_Share
FROM agg_user_table
GROUP BY "Device_Brand"
ORDER BY Total_Users DESC
LIMIT 10;"""
popular_device_brands_df = pd.read_sql(query, engine)
popular_device_brands_df
```

```python
#5.6 Top 10 States with High User Engagement (App Opens per Registered User)
#Insight: Identifies states with high user engagement, measured by app opens per registered user, helping to target retention efforts.
query="""SELECT 
    "State",
    SUM("RegisteredUsers") AS Total_Registered_Users,
    SUM("AppOpens") AS Total_App_Opens,
    ROUND(SUM("AppOpens")::numeric / NULLIF(SUM("RegisteredUsers"), 0), 2) AS App_Open_Rate
FROM map_user_table
GROUP BY "State"
HAVING SUM("RegisteredUsers") > 100000  -- Optional threshold for meaningful data
ORDER BY App_Open_Rate DESC
LIMIT 10;"""
high_engagement_states_df = pd.read_sql(query, engine)
high_engagement_states_df
```
